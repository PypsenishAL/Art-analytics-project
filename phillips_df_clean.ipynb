```python
import pandas as pd
import glob

# Find all phillips_lots_*.csv files
csv_files = glob.glob('phillips_lots_*.csv')
print(f"Found {len(csv_files)} files to concatenate:")
for f in sorted(csv_files):
    print(f"  - {f}")

# Read and concatenate all files
all_dfs = []
for file in sorted(csv_files):
    df = pd.read_csv(file)
    all_dfs.append(df)
    print(f" Loaded {file}: {len(df)} rows")

# Concatenate all DataFrames
if all_dfs:
    combined_df = pd.concat(all_dfs, ignore_index=True)
    
    output_file = 'phillips_all_lots.csv'
    combined_df.to_csv(output_file, index=False, encoding='utf-8-sig')
    print(f"\n Saved combined file: {output_file}")
    print(f" Total rows: {len(combined_df)}")
```

```python
import re

df_cleaned = df[df['price'].notna()].reset_index(drop=True)

def split_estimate(val):
    text = str(val).strip()
    
    if not text or text == 'nan':
        return pd.Series([None, None])
    

    matches = re.findall(r'\d{1,3}(?:,\d{3})*(?:\.\d+)?|\d+(?:\.\d+)?', text)
    
    clean_nums = []
    for m in matches:
        try:
            num = float(m.replace(',', ''))
            clean_nums.append(num)
        except ValueError:
            continue
    
    if len(clean_nums) < 2:
        return pd.Series([None, None])
    
    return pd.Series([clean_nums[0], clean_nums[1]])

df_cleaned[['estimate_low', 'estimate_high']] = df_cleaned['estimate'].apply(split_estimate)
```

```python
def get_currency_code(row):
    curr = row['currency']
    url = str(row['auction_url'])
    
    # Если валюта уже определена как фунт
    if curr == '£':
        return 'GBP'
    
    # Если доллар ($), проверяем локацию
    if curr == '$' or pd.isna(curr):
        if '/HK' in url or 'HONG KONG' in str(row['auction_title']).upper():
            return 'HKD'
        else:
            # По умолчанию считаем USD 
            return 'USD'
            
    return 'USD' 

# Применяем функцию
df_cleaned['currency_code'] = df_cleaned.apply(get_currency_code, axis=1)

# Проверка распределения
print("Распределение валют после обработки:")
print(df_cleaned['currency_code'].value_counts())
```

```python
import pandas as pd
from dateutil import parser as date_parser

def clean_date(val):
    if pd.isna(val) or str(val).strip() == '' or str(val).strip().lower() == 'unknown':
        return pd.NaT
    
    try:
        parsed_date = date_parser.parse(str(val), fuzzy=True)
        return parsed_date.strftime('%Y-%m-%d')
    except Exception:
        # Если не удалось распарсить
        return pd.NaT

# Применяем функцию
df_cleaned['auction_date_clean'] = df_cleaned['auction_date'].apply(clean_date)

# Преобразуем в тип datetime (для удобной сортировки и фильтрации)
df_cleaned['auction_date_clean'] = pd.to_datetime(df_cleaned['auction_date_clean'], errors='coerce')
```

```python
import pandas as pd
from datetime import datetime
from forex_python.converter import CurrencyRates

# Инициализация конвертера
c = CurrencyRates()

df_cleaned['date_obj'] = pd.to_datetime(df_cleaned['auction_date_clean'], errors='coerce')
df_cleaned['year'] = df_cleaned['date_obj'].dt.year
df_cleaned['month'] = df_cleaned['date_obj'].dt.month

#Определяем уникальные пары (Валюта, Год, Месяц), которые нужно запросить
unique_combinations = df_cleaned.dropna(subset=['currency_code', 'year', 'month'])[['currency_code', 'year', 'month']].drop_duplicates()

exchange_rates_map = {}

print(f"Получение курсов валют для {len(unique_combinations)} уникальных комбинаций...")

for _, row in unique_combinations.iterrows():
    curr = row['currency_code']
    year = int(row['year'])
    month = int(row['month'])
    
    if curr == 'USD':
        exchange_rates_map[(curr, year, month)] = 1.0
        continue
    
    try:
        date_val = datetime(year, month, 28)
        rate = c.get_rate(curr, 'USD', date_val)
        exchange_rates_map[(curr, year, month)] = rate
    except Exception:
        try:
            rate = c.get_rate(curr, 'USD')
            exchange_rates_map[(curr, year, month)] = rate
        except:
            exchange_rates_map[(curr, year, month)] = None

#Функция для быстрого расчета
def calculate_final_values(row):
    price = row['price']
    curr = row['currency_code']
    year = row['year']
    month = row['month']
    
    if pd.isna(price) or price == 0 or pd.isna(curr):
        return pd.Series([None, None, None])
    
    rate = exchange_rates_map.get((curr, year, month))
    
    if rate is None:
        return pd.Series([None, None, None])
    
    # Конвертация
    price_usd_raw = price * rate
    
    # Инфляция
    if pd.notna(year):
        years_diff = 2026 - year
        if years_diff > 0:
            price_usd_2026 = price_usd_raw * ((1.035) ** years_diff)
        else:
            price_usd_2026 = price_usd_raw
    else:
        price_usd_2026 = price_usd_raw
        
    return pd.Series([price_usd_raw, price_usd_2026, rate])

#Применяем функцию к df_cleaned
results = df_cleaned.apply(calculate_final_values, axis=1)

df_cleaned['price_usd_raw'] = results[0]
df_cleaned['price_usd_2026'] = results[1]
df_cleaned['applied_rate'] = results[2]

#То же самое для эстимейтов
def calc_estimate(row, col_name):
    val = row[col_name]
    curr = row['currency_code']
    year = row['year']
    month = row['month']
    
    if pd.isna(val) or val == 0 or pd.isna(curr):
        return None
    
    rate = exchange_rates_map.get((curr, year, month))
    if rate is None: return None
    
    raw = val * rate
    if pd.notna(year) and (2026 - year) > 0:
        return raw * ((1.035) ** (2026 - year))
    return raw

if 'estimate_low' in df_cleaned.columns:
    df_cleaned['estimate_low_usd_2026'] = df_cleaned.apply(lambda r: calc_estimate(r, 'estimate_low'), axis=1)

if 'estimate_high' in df_cleaned.columns:
    df_cleaned['estimate_high_usd_2026'] = df_cleaned.apply(lambda r: calc_estimate(r, 'estimate_high'), axis=1)

print("Готово!")
print(df_cleaned[['price', 'currency_code', 'price_usd_raw', 'price_usd_2026']].head(10))
```
