```python
IMAGES_DIR = Path('/content/drive/MyDrive/phillips_lot_images')
IMAGES_DIR.mkdir(exist_ok=True, parents=True)
print(f'Папка создана: {IMAGES_DIR}')
```

```python
df_phillips = pd.read_csv('/content/phillips_cleaned.csv')

print(f'Размерность исходного датасета: {df_phillips.shape}')
df_phillips.head(3)
```

```python
df_phillips = df_phillips.dropna(subset=['image_url', 'price_usd_2026', 'year', 'month']).copy()
df_phillips = df_phillips.drop_duplicates(subset=['lot_url']).reset_index(drop=True)

print(f'Размерность после фильтрации: {df_phillips.shape}')
```

```python
def build_local_path(row) -> str:
    lot_id = str(row['lot_url']).rstrip('/').split('/')[-1]
    ext = os.path.splitext(row['image_url'])[-1]
    if not ext or len(ext) > 5:
        ext = '.jpg'
    return str(IMAGES_DIR / f'{lot_id}{ext}')

df_phillips['local_image_path'] = df_phillips.apply(build_local_path, axis=1)
df_phillips[['lot_url', 'image_url', 'local_image_path']].head(3)
```

```python
ua = UserAgent()

def download_image(url: str, local_path: str, n_retries: int = 3, timeout: int = 10) -> bool:
    '''Скачивает картинку по url и сохраняет в local_path. Возвращает True при успехе.'''
    if os.path.exists(local_path):
        return True

    for attempt in range(n_retries):
        try:
            headers = {'User-Agent': ua.random}
            response = requests.get(url, headers=headers, timeout=timeout)
            response.raise_for_status()

            with open(local_path, 'wb') as f:
                f.write(response.content)

            # проверяем, что файл действительно открывается как изображение
            Image.open(local_path).convert('RGB')
            return True

        except Exception as error:
            if attempt == n_retries - 1:
                print(f'Не удалось скачать {url}: {error}')
            else:
                sleep(0.5 * (attempt + 1))

    return False
```

```python
download_results = []

for _, row in tqdm(df_phillips.iterrows(), total=len(df_phillips), desc='Скачивание изображений'):
    success = download_image(row['image_url'], row['local_image_path'])
    download_results.append(success)

df_phillips['image_downloaded'] = download_results

print(f"Успешно скачано: {sum(download_results)} из {len(download_results)}")
```
